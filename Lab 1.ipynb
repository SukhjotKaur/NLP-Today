# Part 1: Maual Tagging
## Manual labelling done by hand
These give us a reference for comparing the pos annotation results obtained from different models later.
The manually annotated examples are from **general domain**, and it is observed that there can be ambiguity about certain words in these sentences.

!["manually annotated general domain sentences"](Manual_annotation.jpeg)
### Ambiguous Case
The word **flies**:
- In “Time flies”, **flies** is a **verb**.
- In “fruit flies”, **flies** is part of the composite noun and refers to insects.<br>

The word **like**
- In "Time flies like an arrow", **like** is a preposition indicating to a noun (an arrow)
- In "Fruit flies like banana", **like** is used as a verb.
>It shows why PoS tagging cannot always be done reliably by looking at one word in isolation, as the meaning of the owrd is context dependent.


# Part 2: Automatic analysis
## PoS tagging with spaCy

```python
import spacy
import scispacy
from cyk import CYK
import benepar
from nltk import Tree


import warnings

warnings.filterwarnings(
    "ignore",
    message=r".*TreeCRF.*does not define `arg_constraints`.*"
)
```

For the automatic analysis, we use **spaCy** with its English model: en_core_web_sm.

```python
nlp = spacy.load('en_core_web_sm')
```

#### Results of SpaCy for the sentences in the general Domain

```python
gen_sentence = ["Time flies like an arrow; fruit flies like banana.", 
                "The cat sat on the couch.", 
                "The spy saw the cop with the telescope.",
                "The spy saw the cop with the revolver."]

for s in gen_sentence:
  doc = nlp(s)
  print(doc)
  print("-"*40)
  for token in doc:
      print(token.text, token.pos_, token.tag_, spacy.explain(token.tag_))
  print("="*40)
```

### Comparison of Manual and SpaCy annotations:
It can be observed that the spaCy output is more detailed than the short manual annotation. It gives not only a broad PoS category, but also information such as *singular/plural form, verb tense, punctuation type* in the general domain.

However, the manual analysis has an important advantage in ambiguous cases: a human uses the whole context of the sentence to decide which part of speach are the words in the sentence. For example, in **“Time flies like an arrow; fruit flies like banana”**, the first *flies* is a verb, while the second *flies* is a noun. The word *like* also changes its grammatical role across the two parts. On the other hand, the spaCy output captures some of these distinctions (it correctly identifies *flies* in both cases, but fails at *like*), but it does not fully resolve the intended interpretation in the same way a human reader can.

## Part 3: Constituency Parsing

## Trying different grammars for Constituency Parsing using CYK file provided:

We are trying to do the results for a grammar wiht lesser number of rules

```python
G1 = [
        #S is axiom
        ("S", ("NP", "VP")), 
                    
                    # non terminal rules
                    ("NP", ("DET", "NOUN")),
                    ("PP", ("ADP", "NP")),
                    ("VP", ("VERB", "PP")),
                    
                    # terminal (if using pos directly)
                    ("DET", ("DET",)),
                    ("VERB", ("VERB",)),
                    ("NOUN", ("NOUN",)),
                    ("ADP", ("ADP",))
                ]

cyk1 = CYK(G1)
```

```python
sentence1 = ["Time", "flies", "like", "an", "arrow"]

pos1 = [["NOUN"], ["VERB", "NOUN"], ["ADP", "VERB"], ["DET"], ["NOUN"]]

trees1 = cyk1(pos1, sentence1)

sentence_trees1 = [
    tree for tree in trees1
    if tree.type == "S"
]

print("Number of trees:", len(sentence_trees1))

for tree in sentence_trees1:
    print(tree)
```

```python
sentence2 = ["Fruit", "flies", "like", "a", "banana"]

pos2 = [["NOUN"], ["NOUN", "VERB"], ["VERB", "ADP"], ["DET"], ["NOUN"]]

trees2 = cyk1(pos2, sentence2)

sentence_trees2 = [
    tree for tree in trees2
    if tree.type == "S"
]   

print("Number of trees:", len(sentence_trees2))

for tree in sentence_trees2:
    print(tree)
```

```python
sentence3 = ["The", "cat", "sat", "on", "the", "couch"]

pos3 = [["DET"], ["NOUN"], ["VERB"], ["ADP"], ["DET"], ["NOUN"]]

trees3 = cyk1(pos3, sentence3)

sentence_trees3 = [
    tree for tree in trees3
    if tree.type == "S"
]
print("Number of trees:", len(sentence_trees3))

for tree in sentence_trees3:
    print(tree)
```

### Observations with the restricted grammar

The restricted grammar recognises only sentences matching its small set of production rules. It can analyse “The cat sat on the couch” because the sentence follows `S -> NP VP`, with `VP -> VERB PP`. The other sentences require structures (like "The spy saw the cop with the revolver") that are absent from this grammar, such as a `NP -> Noun` or `NP -> NOUN VP` and other such rules which makes the grammar more comprehensive. 
>Therefore, the absence of a tree indicates limited grammar coverage rather than an ungrammatical sentence.

#### Now trying the results on a more comprehensive grammer with more rules

```python
G2 = [
    # Sentence
    ("S", ("NP", "VP")),

    # Noun phrases
    ("NP", ("NP", "PP")),
    ("NP", ("NOUN", "NOUN")),
    ("NP", ("DET", "NOUN")),
    ("NP", ("NP", "NOUN")),
    ("NP", ("ADJ", "NOUN")),
    ("NP", ("ADJ", "NP")),
    ("NP", ("PRON",)),
    ("NP", ("PROPN",)),
    ("NP", ("NP", "VP")),

    # Prepositional and clause adjuncts
    ("PP", ("ADP", "NP")),
    ("PP", ("SCONJ", "S")),
    ("PP", ("PRON", "VP")),

    # Verb phrases
    ("VP", ("VP", "PP")),
    ("VP", ("VERB", "PP")),
    ("VP", ("VERB", "NP")),
    ("VP", ("AUX", "VP")),
    ("VP", ("AUX", "NP")),

    # Terminal rules
    ("NP", ("NOUN",)),
    ("NOUN", ("NOUN",)),
    ("VERB", ("VERB",)),
    ("ADP", ("ADP",)),
    ("DET", ("DET",)),
    ("ADJ", ("ADJ",)),
    ("PRON", ("PRON",)),
    ("AUX", ("AUX",)),
    ("SCONJ", ("SCONJ",)),
    ("PROPN", ("PROPN",)),
]

cyk2 = CYK(G2)
```

```python
sentence1 = ["Time", "flies", "like", "an", "arrow"]

pos1 = [["NOUN"], ["VERB", "NOUN"], ["ADP", "VERB"], ["DET"], ["NOUN"]]

trees1 = cyk2(pos1, sentence1)
sentence_trees1 = [
    tree for tree in trees1
    if tree.type == "S"
]

print("Number of trees:", len(sentence_trees1))

for tree in sentence_trees1:
    print(tree)
```

```python
sentence2 = ["Fruit", "flies", "like", "a", "banana"]

pos2 = [["NOUN"], ["NOUN", "VERB"], ["VERB", "ADP"], ["DET"], ["NOUN"]]

trees2 = cyk2(pos2, sentence2)

sentence_trees2 = [
    tree for tree in trees2
    if tree.type == "S"
]

print("Number of trees:", len(sentence_trees2))

for tree in sentence_trees2:
    print(tree)
```

```python
sentence3 = ["The", "cat", "sat", "on", "the", "couch"]

pos3 = [["DET"], ["NOUN"], ["VERB"], ["ADP"], ["DET"], ["NOUN"]]

trees3 = cyk2(pos3, sentence3)

sentence_trees3 = [
    tree for tree in trees3
    if tree.type == "S"
]

print("Number of trees:", len(sentence_trees3))

for tree in sentence_trees3:
    print(tree)
```

```python
sentence4 = ["The", "spy", "saw", "the", "cop", "with", "the", "telescope"]

pos4 = [["DET"], ["NOUN"], ["VERB"], ["DET"], ["NOUN"], ["ADP"], ["DET"], ["NOUN"]]

trees4 = cyk2(pos4, sentence4)

sentence_trees4 = [
    tree for tree in trees4
    if tree.type == "S"
]

print("Number of trees:", len(sentence_trees4))

for tree in sentence_trees4:
    print(tree)
```

```python
sentence5 = ["The", "spy", "saw", "the", "cop", "with", "the", "revolver"]

pos5 = [["DET"], ["NOUN"], ["VERB"], ["DET"], ["NOUN"], ["ADP"], ["DET"], ["NOUN"]]

trees5 = cyk2(pos5, sentence5)

sentence_trees5 = [
    tree for tree in trees5
    if tree.type == "S"
]

print("Number of trees:", len(sentence_trees5))

for tree in sentence_trees5:
    print(tree)
```

### Results of the extended grammar

The extended grammar produces more analyses because it includes additional rules such as `VP → VERB NP`, `NP → NP PP`, and `VP → VP PP`. These rules allow it to analyse sentences whihch follow more comprehensive rules.

For “Time flies like an arrow”, assigning multiple possible PoS tags to *flies* and *like* produces different interpretations. In the intended interpretation, *flies* is a verb and *like* is a adverb-preposition. The added rules also permit a reading in which *flies* is a noun and *like* is a verb. This demonstrates how the selected PoS tags affect the resulting constituency trees.

The telescope and revolver sentences each produce two trees. In one tree, the phrase beginning with *with* modifies “the cop”; in the other, it modifies the action “saw”. This ambiguity is produced by differnt verb phrase possiblilities.


### Comparison of the results of the two grammars
It is observed that in the CYK experiments, parsing depends on both the possible parts of speech tags and the grammar rules supplied. A restricted grammar has limited coverage (we see some sentences have 0 trees), while a more comprehensive grammar recognises more sentences but may also produce multiple valid trees.

## CYK analysis of the scientific-domain sentences

The scientific sentences contain structures not covered by the previous grammar, including adjective sequences, auxiliaries, subordinate clauses and relative clauses. We therefore extend the grammar while retaining the principal phrase categories `NP`, `VP` and `PP`.

Punctuation is omitted because it does not affect the phrase structures examined here and no punctuation productions are included in the grammar.

For the scientific texts we use the more comprehensive grammar to see its constituency parsing

```python
biology = [ "Arabidopsis", "thaliana", "seedlings", "exhibit", "longer", "hypocotyls",
           "when", "they", "are", "grown", "under", "high", "ambient", "temperature",  
           "which", "is", "defined", "as", "thermomorphogenesis"]

biology_pos = [["NOUN"], ["NOUN"], ["NOUN"], ["VERB"], ["ADJ"], ["NOUN"], ["SCONJ"], ["PRON"], ["AUX"],
               ["VERB"], ["ADP"], ["ADJ"], ["ADJ"], ["NOUN"], ["PRON"], ["AUX"], ["VERB"], ["ADP"], ["NOUN"]]

trees = cyk2(biology_pos, biology)

sentence_trees = [
    tree for tree in trees
    if tree.type == "S"
]

print("Number of sentence trees:", len(sentence_trees))

for tree in sentence_trees:
    print(tree)
```

```python
astronomy = ["A", "spectrogram", "of", "PSN J10354824+3900279", "obtained", "on", "Dec. 19.33 UT", "suggests",
             "that", "this", "is", "a", "type-Ia", "at", "redshift", "z 0.044"]

astronomy_pos = [["DET"], ["NOUN"], ["ADP"], ["PROPN"], ["VERB"], ["ADP"], ["PROPN"], ["VERB"],
                 ["SCONJ"], ["PRON"], ["AUX"], ["DET"], ["NOUN"], ["ADP"], ["NOUN"], ["NOUN"]]

astronomy_results = cyk2(astronomy_pos, astronomy)

astronomy_trees = [
    tree for tree in astronomy_results
    if tree.type == "S"
]

print(
    "Number of complete sentence trees:",
    len(astronomy_trees)
)

# Display only the first few to keep the report readable
for tree in astronomy_trees[:3]:
    print(tree)
```

Compared to the general-domain examples, the scientific texts required additional rules for adjectives, pronouns, auxilary verbs, proper nouns and clauses. This demonstrates that the covering power of a constituency parser depends on its grammar containing the structures and PoS categories found in the target domain.

## Constituency parsing using the Berkeley Neural Parser

The same general- and scientific-domain sentences are analysed using the Berkeley Neural Parser through its spaCy integration. The resulting bracketed trees show the constituent labels and their hierarchical relationships. These results are subsequently compared with the trees generated by the rule-based CYK parser.

```python
nlp_constituency = spacy.load("en_core_web_sm")

nlp_constituency.add_pipe("benepar", config={"model": "benepar_en3"})
```

```python
print(nlp_constituency.pipe_names)
```

```python
for text in gen_sentence:
    doc = nlp_constituency(text)

    for sent in doc.sents:
        print("\nSentence:", sent.text)
        print(sent._.parse_string)

        tree = Tree.fromstring(sent._.parse_string)
        tree.pretty_print()
```

```python
for text in sci_sentence:
    doc = nlp_constituency(text)

    for sent in doc.sents:
        print("\nSentence:", sent.text)
        print(sent._.parse_string)

        tree = Tree.fromstring(sent._.parse_string)
        tree.pretty_print()
```

### Comparison of CYK and the Berkeley Neural Parser

The CYK parser and the Berkeley Neural Parser both produce constituency structures, but they follow different approaches. 
- CYK generates every structure permitted by the manually supplied PoS alternatives and grammar rules. 
- The Berkeley parser uses a pretrained model and returns one ranked analysis with a
larger collection of constituent labels.

#### General-domain sentences

For “Time flies like an arrow”, CYK retains different interpretations of *flies* and *like*, while the Berkeley parser selects the intended reading with *flies* as the verb and “like an arrow” as a prepositional phrase.<br>
Both parsers produce the same main structure for “The cat sat on the couch”. For the two spy sentences, CYK produces both possible attachments of the phrase beginning with *with*. The Berkeley parser selects its attachment to the verb phrase in both cases.

#### Scientific-domain sentences

The scientific sentences required additional CYK rules for adjectives, pronouns, auxiliaries, clauses etc. This increased the number of possible CYK trees. The Berkeley parser represents the scientific sentences using more detailed constituents, including subordinate clauses, and returns one selected parse.<br>
Therefore, CYK makes the effects of grammar rules and ambiguity visible, whereas the Berkeley parser provides broader coverage and a single contextual analysis.

## Observations

The experiments show that constituency parsing results depend on both PoS interpretation and grammar coverage. The restricted CYK grammar is transparent and produces understandable results for simple constructions, but valid sentences receive no tree when a necessary rule is absent. Extending the grammar
improves coverage but also increases the number of possible analyses.

CYK preserves all interpretations endorsed by the supplied grammar. This is useful for observing lexical and structural ambiguity, as demonstrated by *flies*, *like*, and the two possible attachments of the phrases beginning with *with*.

The Berkeley Neural Parser uses a much richer set of constituent parts of speech and returns one selected analysis. It handles the long scientific sentences more naturally, including subordinate and relative clauses, although the fruit-flies example shows that a pretrained parser can still choose an unsuitable interpretation. Therefore, the hand-written CYK parser offers transparency, while the pretrained parser provides broader coverage and contextual selection.
